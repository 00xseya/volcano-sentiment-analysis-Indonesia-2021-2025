```python
import pandas as pd
from pathlib import Path
```

```python
SOURCE_DATA = Path("../../data/ner_distribution/data_text.csv")
OUTPUT_DIR = Path("../../random_process/output")
# OUTPUT_FILE = OUTPUT_DIR / "splited_data.csv"
# OUTPUT_FILE.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(SOURCE_DATA)
print(f"Original rows: {len(df):,}\n")
```

```python
print("Year distribution in original data: ")
print(df["video_year"].value_counts().sort_index())
print()
```

### Only take 30% data

```python
data_sample = (
    df.groupby("video_year", group_keys=False).apply(lambda g: g.sample(frac=0.30, random_state=42)).reset_index(drop=True)
)
```

```python
print(f"Output 30% sample: {len(data_sample):,} rows")
```

### Remaining 70% data

```python
remain_sample = df.drop(data_sample.index).reset_index(drop=True)
```

```python
print(f"Output remaining 70% sample: {len(remain_sample):,} rows")
```
