```python
DATA_TEXT_PATH = "../data/ner_distribution/data_text.csv"
```

```python
df = pd.read_csv(DATA_TEXT_PATH) 
output_data = []
```

```python
for index, row in df.iterrows():
    text = str(row['text_cleansed'])
    text_lower = text.lower()
    
    annotations = []
    
    for category, phrases in PHRASES.items():
        label = ENTITY_TYPES[category]
        for phrase in phrases:
            start_idx = 0
            while True:
                idx = text_lower.find(phrase, start_idx)
                if idx == -1:
                    break
                
                end_idx = idx + len(phrase)
                
                annotations.append({
                    "from_name": "label",
                    "to_name": "text",
                    "type": "labels",
                    "value": {
                        "start": idx,
                        "end": end_idx,
                        "text": text[idx:end_idx],
                        "labels": [label]
                    }
                })
                start_idx = end_idx 

    task = {
        "data": {
            "text": text,
            "video_title": str(row.get('video_title', '')),
            "video_year": str(row.get('video_year', '')),
        },
        "annotations": [{
            "result": annotations
        }]
    }
    output_data.append(task)
```

```python
with open("../data/ner_distribution/pre_annotated_data.json", "w", encoding="utf-8") as f:
    json.dump(output_data, f, indent=2, ensure_ascii=False)

print(f"Generated pre-annotations for {len(output_data)} sentences.")
```
